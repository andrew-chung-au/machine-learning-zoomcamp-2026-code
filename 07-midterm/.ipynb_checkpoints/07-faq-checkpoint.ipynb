```python
# Imports and environment

import os
import sys
import sklearn
import scipy
import pandas as pd

from scipy.io import arff
from sklearn.model_selection import GridSearchCV, cross_validate,train_test_split, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

print("Python:", sys.version)
print("Context-aware warnings:", sys.flags.context_aware_warnings)
print("scikit-learn:", sklearn.__version__)
print("SciPy:", scipy.__version__)
```

```python
# Load and prepare the original dataset
raw, meta = arff.loadarff("training-dataset.arff")
df = pd.DataFrame(raw)
df = df.apply(lambda s: s.str.decode("utf-8")).astype(int)
df.columns = df.columns.str.lower()

df["phishing"] = (df["result"] == -1).astype(int)
df = df.drop(columns=["result"])

# Split the data, preserving the original settings
SEED = 1

df_full_train, df_test = train_test_split(
    df, test_size=0.2, random_state=SEED, stratify=df.phishing
)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=SEED,
    stratify=df_full_train.phishing
)

# Define the variables used in the diagnostic
features = [col for col in df.columns if col != "phishing"]
X_full = df_full_train[features]
y_full = df_full_train["phishing"]

print("X_full:", X_full.shape)
print("y_full:", y_full.shape)
```

```python
# Cross-validation and model configurations

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=SEED,
)

searches = {
    "Logistic regression (one-hot)": (
        make_pipeline(
            OneHotEncoder(handle_unknown="ignore"),
            LogisticRegression(max_iter=2000),
        ),
        {"logisticregression__C": [0.01, 0.1, 1, 10, 100]},
    ),
    "Decision tree": (
        DecisionTreeClassifier(random_state=SEED),
        {
            "max_depth": [3, 5, 8, 12, None],
            "min_samples_leaf": [1, 5, 10, 20],
        },
    ),
    "Random forest": (
        RandomForestClassifier(random_state=SEED, n_jobs=1),
        {
            "n_estimators": [200],
            "max_depth": [10, 20, None],
            "min_samples_leaf": [1, 3, 5],
            "max_features": ["sqrt", 0.5],
        },
    ),
    "XGBoost": (
        XGBClassifier(
            random_state=SEED,
            eval_metric="logloss",
            n_jobs=-1,
        ),
        {
            "n_estimators": [200, 400],
            "max_depth": [3, 5, 8],
            "learning_rate": [0.05, 0.1, 0.3],
        },
    ),
}
```

```python
# Run the complete original workflow

best, summ = {}, []

for name, (est, grid) in searches.items():
    print(f"\n--- {name}: GridSearchCV.fit() ---", flush=True)

    gs = GridSearchCV(
        est,
        grid,
        scoring="roc_auc",
        cv=cv,
        n_jobs=-1,
    )
    gs.fit(X_full, y_full)

    best[name] = gs.best_estimator_

    print(f"--- {name}: cross_validate() ---", flush=True)

    r = cross_validate(
        gs.best_estimator_,
        X_full,
        y_full,
        cv=cv,
        scoring=["roc_auc", "precision", "recall", "f1"],
    )

    summ.append({
        "model": name,
        "best_params": gs.best_params_,
        "roc_auc": r["test_roc_auc"].mean(),
        "precision": r["test_precision"].mean(),
        "recall": r["test_recall"].mean(),
        "f1": r["test_f1"].mean(),
    })

print("Test completed.")
```

```python
# Step 1 — Run this minimal test (Add a temporary new cell)
import warnings
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

X, y = make_classification(
    n_samples=500,
    n_features=10,
    random_state=SEED,
)

cv_small = StratifiedKFold(n_splits=2, shuffle=True, random_state=SEED)

# Test outer and inner parallelism independently.
for grid_jobs, forest_jobs in [(1, 1), (1, -1), (-1, 1), (-1, -1)]:
    print(
        f"\nGridSearchCV n_jobs={grid_jobs}, "
        f"RandomForest n_jobs={forest_jobs}",
        flush=True,
    )

    gs = GridSearchCV(
        RandomForestClassifier(
            random_state=SEED,
            n_jobs=forest_jobs,
        ),
        {"n_estimators": [10], "max_depth": [5]},
        scoring="roc_auc",
        cv=cv_small,
        n_jobs=grid_jobs,
    )

    with warnings.catch_warnings():
        warnings.filterwarnings(
            "always",
            message=r"`sklearn\.utils\.parallel\.delayed` should be used with",
            category=UserWarning,
        )
        gs.fit(X, y)

    print("Completed", flush=True)
```

```python
import warnings
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

_, full_grid = searches["Random forest"]

test_grids = {
    "One combination": {
        "n_estimators": [200],
        "max_depth": [10],
        "min_samples_leaf": [1],
        "max_features": ["sqrt"],
    },
    "Six combinations": {
        "n_estimators": [200],
        "max_depth": [10, 20, None],
        "min_samples_leaf": [1],
        "max_features": ["sqrt", 0.5],
    },
    "Full grid": full_grid,
}

for name, grid in test_grids.items():
    print(f"\n--- {name} ---", flush=True)

    gs = GridSearchCV(
        RandomForestClassifier(
            random_state=SEED,
            n_jobs=-1,
        ),
        grid,
        scoring="roc_auc",
        cv=cv,
        n_jobs=-1,
    )

    with warnings.catch_warnings():
        warnings.filterwarnings(
            "always",
            message=r"`sklearn\.utils\.parallel\.delayed` should be used with",
            category=UserWarning,
        )
        gs.fit(X_full, y_full)

    print("Completed", flush=True)
```
